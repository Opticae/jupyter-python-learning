```python
# Concept 1: The 'for' Loop
# The syntax

fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print (f"I love {fruit}!")

# -- how this works --
# 1. python looks at the list 'fruits'
# 2. it takes the first item ("apple") and assigns to the var 'fruit'
# 3. it runs the code inside the loop (print)
# 4. it goes back to the top, grabs the next item, assigns it to fruit and runs the
#    the code again
# 5. the code stops automatically when it runs out of items - no 'break' needed!
```

```python
# concept 2: The 'range()' Function

# if we dont have a list but want to repeat something for a specific number of times?

# Prints 0, 1, 2, 3, 4 (stops BEFORE 5)
for i in range(5):
    print(i)

# Prints 1, 2, 3, 4, 5 (starts at 1, stops BEFORE 6)
for i in range(1, 6):
    print(i)
```

```python
# concept 3 - functions 'def'

# functions allow you to group code into reusable blocks with a name. Write code
# once and 'call' it whatever you need. - DRY principle (Dont Repeat Yourself).

# 1. DEFINE the function (this just saves it in memory, it doesn't run yet)
def greet_user(name):
    print(f"Hello, {name}! Welcome to the program.")

# 2. CALL the function (this actually runs the code inside it)
greet_user("Alice")
greet_user("Bob")
```

```python
# Assignment 37: The Grocery Receipt (Basic 'for' loop)
# The Problem Statement:
# Write a program that prints a formatted receipt for a grocery list.
# Create a list called groceries: ["Milk", "Eggs", "Bread", "Butter"]
# Print "--- YOUR RECEIPT ---"
# Use a for loop to go through every item in the groceries list.
# Inside the loop, print each item formatted like this: " - [item]"
# After the loop finishes, print "------------------" and "Total items: [number]" (Hint: use len()).

groceries = ["Milk", "Eggs", "Bread", "Butter"]

print("--- YOUR RECEIPT ---")
for grocery in groceries:
    print(f"- {grocery}")
print("------------------")
print(f"Total items: {len(groceries)}")
```

```python
# Assignment 38: The Number Squarer (range + for)
# The Problem Statement:
# Write a program that asks the user for a number, and then prints the square of every number from 1 up to that number.
# Ask the user: "Enter a number: " (convert to integer).
# Use a for loop with range(1, user_number + 1) to count up to their number.
# Inside the loop, calculate the square (number * number or number ** 2).
# Print: "[number] squared is [square]"

user_number = int(input("Enter a number: "))

for i in range(1, user_number + 1):
    squared = i ** 2
    print(f"{i} squared is {squared}")
```

```python
# Assignment 39: The Menu Display (Combining Functions + for Loops)
# The Problem Statement:
# Let's refactor your To-Do list menu into a clean, reusable function!
# Define a function called display_menu(options). It should take one parameter: a list of strings.
# Inside the function, print "=== MAIN MENU ==="
# Use a for loop to go through the options list.
# Print each option. (Bonus: Try to print them with numbers, like "1. Add task", "2. Remove task". 
# You can use a counter variable that starts at 1 and adds 1 each loop, or look up Python's enumerate() function!)
# Print "=================" at the end of the function.
# Outside the function, create a list: my_options = ["Add a task", "Remove a task", "View tasks", "Quit"]
# Call the function and pass my_options into it: display_menu(my_options)

def display_menu(options):
    print("=== MAIN MENU ===")

    for number, option in enumerate(options, start=1):
        print(f"{number}. {option}")

    print("=================")

options = ["Add a task", "Remove a task", "View task", "Quit"]
display_menu(options)
```
