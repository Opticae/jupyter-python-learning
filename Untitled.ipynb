```python
# Opens a file called "test.txt" in write mode
# Writes "I am learning Python!"
# Then opens the same file in read mode
# Reads the content
# Prints it

file = open("test.txt", "w")
file.write("I am learning in Python!")
file.close()

file = open("test.txt", "r")
content = file.read()
file.close()
print(content)
```
