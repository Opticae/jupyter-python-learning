```python
# Assignment 30: The Dynamic To-Do List

# 1. todo_list - ["Buy groceries", "Walk the dog"]
todo_list = ["Buy groceries", "Walk the dog"]

# 2. print "todo_list"
print(f"Current tasks: {todo_list}")

# 3. query user to (a)dd or (r)emove a task
user_query = input("Do you want to (a)dd a task or (r)emove a task?").strip().lower()

# 4. use if/elif/else to handle query
#     * "a" - "What task do you want to add?" use .append().capitalize()
#         - print("Task added!")
#     * "r" - "What task do you want to remove?" use .remove()
#         - print("Task removed!")
#     else print("That task is not on your list.") 

if user_query == "a":
    add_task = input("What task do you want to add? ").capitalize()
    todo_list.append(add_task)
    print("Task added!")
elif user_query == "r":
    remove_task = input("What task do you want to remove? ").strip().capitalize()
    if remove_task in todo_list:
        todo_list.remove(remove_task)
        print("Task removed!")
    else:
        print("That task is not on your list.")
else:
    print("Invalid choice.")
    
# 5. print todo_list at end to see changes
print(f"Final tasks: {todo_list}")
```

```python
# Assignment 32: The Continuous To-Do List

# 1. todo_list - ["Buy groceries", "Walk the dog"]
todo_list = ["Buy groceries", "Walk the dog"]

# 2. print "todo_list"
print(f"Current tasks: {todo_list}")

# 3. Start continuous loop 
while True: 
    user_query = input("Do you want to (a)dd a task, (r)emove a task or (q)uit").strip().lower()

    if user_query == "q":
        print("Exiting to-do list. Goodbye!")
        break
    elif user_query == "a":
        add_task = input("What task do you want to add? ").capitalize()
        todo_list.append(add_task)
        print("Task added!")
        print(f"Current tasks: {todo_list}")
    elif user_query == "r":
        remove_task = input("What task do you want to remove? ").strip().capitalize()
        if remove_task in todo_list:
            todo_list.remove(remove_task)
            print("Task removed!")
            print(f"Current tasks: {todo_list}")
        else:
            print("That task is not on your list.")
    else:
        print("Invalid choice. Please choose 'a', 'r' or 'q'.")
    
# 5. print todo_list at end to see changes
print(f"Final tasks: {todo_list}") 
```

```python
# Loop - repeats block of code
# for loops and while loops
# for loops - repeat for each item 
#     usage: 
#         for task in todo_list:
#             print(task) 

# while loops - repeat as long as condition is true
#  - use when unsure how many times to repeat
# count = 0
# while count <= 3:
#     print(count)
#     count += 1

# while true - intentional infinite loop
# condition after "while" must = True or false.
#     if true - its always true so loop never ends on its own
#     useful if you want to keep going until you decide its time to stop.
#     the way out of this is "BREAK"

while True:
    answer = input("Enter a number (q to quit): ")
    if answer == "q":
        break
    if not answer.isdigit():
        print("Numbers only!")
        continue            # back to the top, skips the line below
    print("Double:", int(answer) * 2)

# SUMMARY:
# 1. FOR - when you have a collection or a known count.
# 2. WHILE CONDITION - when you repeat until something changes 
# 3. WHILE TRUE + BREAK - when the stopping point depends on what happens
# during the loop like user input
```

```python
# Assignment 34 - Number guessing game (counter + condition)

attempts = 0
secret_number = 7

#start loop to prompt user for the secret number 
while True:
    guess_prompt = int(input("Guess the number (1-10): "))
    #Increment attempts
    attempts += 1
    
    #if guess correctly
    if guess_prompt == secret_number:
        print(f"Correct! You got it in {attempts} tries")
        break

    #if reach 5 attempts and still incorrect, game over
    elif attempts == 5:
        print(f"Game Over! the number was {secret_number}.")
        break

    #if wrong and still have attempts left
    else:
        print(f"Incorrect! you have {5 - attempts} tries left.")
        
```

```python
# Assignment 35: The Running Total (Accumulator Pattern)

# create variable total
total = 0 

# use while true loop
while True:
    add_number = float(input("Enter a number (0 to stop): "))
    
    if add_number == 0:
        print(f"Final total: {total}")
        break
    
    #add input number to total
    else:
        total += add_number
        print(f"Running total: {total}")
    
#initially i had the impression that i have to add the number to the total
# at the start of the loop. But running through the data flow i realized that
# it didnt make sense if the break exists where the user adds the number thats
# when i pivoted and corrected the mistake. This is where i started to understand
# how the while loop works and what needs to laid out in the beginning.
```
