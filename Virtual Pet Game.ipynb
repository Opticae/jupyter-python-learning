```python
# 1. Ask the use for their pet's name
name = input("What is your pet's name? ").capitalize()

# 2. Setup starting stats (0-100 Scale)
hunger = 50
happiness = 50

print(f"\nWelcome! You are now taking care of {name}.")

# 3. game loop

while True:
    #print current status
    print(f"\n=== {name}'S STATUS ===")
    print(f"Hunger: {hunger} | Happines: {happiness}")

    #user choice
    choice = input("\nWhat do you want to do? (f)eed, (p)lay, (s)tats, or (q)uit? ").strip().lower()

    #handle user choice
    if choice == "f":
        hunger = max(0, hunger - 20)
        print(f"You fed {name}! Yum!")

    elif choice == "p":
        happiness = min(100, happiness + 15)
        hunger = min(100, hunger + 10)
        print(f"You played with {name}! So fun!")

    elif choice == "s":
        print(f"{name} is doing okay.")

    elif choice == "q":
        print(f"{name} hopes to see you soon! Goodbye!")
        break

    else:
        print("Invalid Choice. Please type f, p, s, or q")

    hunger = min(100, hunger + 5)
    happiness = max(0, happiness - 5)

    # 6. Check for Game Over conditions
    
    # Rule 1: Is the pet starving?
    if hunger >= 100:
        print(f"Game Over! {name} is too hungry! 😢")
        break
    
    # Rule 2: Is the pet too sad?
    if happiness <= 0:
        print(f"Game Over! {name} is too sad! 😢")
        break

print(f"Thanks for playing with {name}!")
        
```
