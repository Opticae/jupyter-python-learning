```python
# Week 3: Persistence & Polish

# concept 1 - file i/o (read, write)
# week 2, lists only exist while program is running. 
# "File I/O" saves data to a text file

# WRITING TO A FILE

# 1. Open a file in "write" mode 
# "w" is write mode. "with open(..) as file:" is the safest way to open files
with open("my_file.txt", "w") as file: # if file doesnt exist, Python creates it
    file.write("Hello, World!\n") #"\n" new line character, like pressing enter
    file.write("This is line 2.\n")   # if file does exist, Overwrites it

# "a" means "append mode" (adds to the end, doesn't overwrite)

with open("my_file.txt", "a") as file:
    file.write("This is a new line at the end.\n")

# "r" means "read mode"

with open("my_file.txt", "r") as file:
    content = file.read()  # Reads the entire file as one string
    print(content)

# to read line by line
with open("my_file.txt", "r") as file:
    for line in file:
        print(line.strip())  # strip() removes the \n at the end
```

```python
# concept 2: Saving and loading lists

# convert list to a string, save to file and load it back later 

# Saving a list:
tasks = ["Buy milk", "Walk dog", "Do laundry"] 

with open("tasks.txt", "w") as file: # Converting list to string and storing it
    for task in tasks:
        file.write(task + "\n")

with open("tasks.txt", "r") as file: 
    for line in file:
        tasks.append(line.strip())
```

```python
# Assignment 40: The Persistent To-Do List (File I/O)

# The Problem Statement:
# Modify your To-Do list to save tasks to a file and load them when the program starts.
# At the start of the program, try to load tasks from "tasks.txt"
# If the file exists, read each line and add it to a todo_list
# If the file doesn't exist (first time running), start with an empty list
# Use your existing while True loop for the menu
# When the user adds a task, also append it to the file
# When the user removes a task, rewrite the entire file with the updated list
# When they quit, print "Goodbye!"
FILENAME = "tasks.txt"

todo_list = []

try:
    with open(FILENAME, "r") as file:
        for line in file:
            task = line.strip() # to remove the new line at the end

            if task: #skip blank lines
                todo_list.append(task)

except FileNotFoundError:
    pass # first time running: the file doesnt exist, so start with an empty list

def save_all_tasks():
    """rewrite the whoel file with the current list"""
    with open(FILENAME, "w") as file:
        for task in todo_list:
            file.write(task + "\n")

#--- Menu loop ---

while True:
    print("\n--- To-Do List ---")
    print("1. View tasks")
    print("2. Add a task")
    print("3. Remove a task")
    print("4. Quit")

    choice = input("Choose an option (1-4): ")
    
    if choice == "1":
        if len(todo_list) == 0:
            print("No tasks yet.")
        else:
            for number, task in enumerate(todo_list, start=1):
                print(f"{number}. {task}")
 
    elif choice == "2":
        task = input("Enter a new task: ").strip()
        if task:
            todo_list.append(task)
            # Append just this one task to the file
            with open(FILENAME, "a") as file:
                file.write(task + "\n")
            print("Task added.")
        else:
            print("Task can't be empty.")
 
    elif choice == "3":
        if len(todo_list) == 0:
            print("Nothing to remove.")
        else:
            for number, task in enumerate(todo_list, start=1):
                print(f"{number}. {task}")
            try:
                number = int(input("Number of the task to remove: "))
                if 1 <= number <= len(todo_list):
                    removed = todo_list.pop(number - 1)
                    save_all_tasks()   # rewrite the file with the updated list
                    print(f"Removed: {removed}")
                else:
                    print("That number isn't on the list.")
            except ValueError:
                print("Please enter a number.")
 
    elif choice == "4":
        print("Goodbye!")
        break
 
    else:
        print("Invalid choice. Please enter 1, 2, 3 or 4.")


```
